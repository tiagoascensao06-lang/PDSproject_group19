MISSING STREAMS

```python
artist_streams.head()
```

```python
cols = ['Total Streams', 'Lead Streams', 'Feature Streams', 'Solo Streams', 'Collaborative Streams']
complete = artist_streams[(artist_streams[cols].isna() == False).all(axis=1)]
complete

diff_solo = complete['Total Streams'] - (complete['Solo Streams'] + complete['Collaborative Streams'])
diff_lead = complete['Total Streams'] - (complete['Lead Streams'] + complete['Feature Streams'])

print('Total = Solo + Collaborative')
print(f'  largest absolute difference: {diff_solo.abs().max()} millions of stream')
print(f'  largest absolute difference: {(diff_solo.abs().max())} millions of stream')
print(f'  largest relative difference: {((diff_solo.abs() / complete["Total Streams"]).max() *100)} %')
print('Very small numbers')
print('\nTotal = Lead + Feature')
print ( f'  Rows with a difference above 0.01 millions of streams : {(diff_lead.abs() > 0.01).sum()} millions of stream')
print(f'  largest absolute difference: {(diff_lead.abs().max()).round(2)} millions of stream')
print(f'  largest relative difference: {((diff_lead.abs() / complete["Total Streams"]).max() *100).round(2)} %')
```

```python
print('Missing streams per column:')
print(artist_streams[cols].isna().sum())
cols = ['Total Streams', 'Lead Streams', 'Feature Streams', 'Solo Streams', 'Collaborative Streams']

max_miss = int(artist_streams[cols].isna().sum(axis=1).max() )
max_miss

ids_with_missing = artist_streams[artist_streams[cols].isna().any(axis=1)].index
ids_with_missing
```

```python
artist_streams.head()
```

```python
nan_total = artist_streams['Total Streams'].isna()
artist_streams.loc[nan_total, 'Total Streams'] = artist_streams.loc[nan_total, 'Solo Streams'] + artist_streams.loc[nan_total, 'Collaborative Streams']

nan_solo = artist_streams['Solo Streams'].isna()
artist_streams.loc[nan_solo, 'Solo Streams'] = artist_streams.loc[nan_solo, 'Total Streams'] - artist_streams.loc[nan_solo, 'Collaborative Streams']

nan_collab = artist_streams['Collaborative Streams'].isna()
artist_streams.loc[nan_collab, 'Collaborative Streams'] = artist_streams.loc[nan_collab, 'Total Streams'] - artist_streams.loc[nan_collab, 'Solo Streams']

nan_lead = artist_streams['Lead Streams'].isna()
artist_streams.loc[nan_lead, 'Lead Streams'] = artist_streams.loc[nan_lead, 'Total Streams'] - artist_streams.loc[nan_lead, 'Feature Streams']

nan_feature = artist_streams['Feature Streams'].isna()
artist_streams.loc[nan_feature, 'Feature Streams'] = artist_streams.loc[nan_feature, 'Total Streams'] - artist_streams.loc[nan_feature, 'Lead Streams']

artist_streams.loc[ids_with_missing, ['Artist Name'] + cols]
```

```python
print(artist_streams[cols].isna().sum())   
```

```python
negative = (artist_streams[cols] < 0).any(axis=1)
artist_streams.loc[negative, ['Artist Name'] + cols]
```

```python
artist_streams.loc[artist_streams['Feature Streams'] < 0, 'Feature Streams'] = 0
artist_streams.loc[artist_streams['Lead Streams'] < 0, 'Lead Streams'] = 0
```

```python
artist_streams.loc[ids_with_missing, ['Artist Name'] + cols]
```

```python
lead_too_big = artist_streams['Lead Streams'] > artist_streams['Total Streams']
feat_too_big = artist_streams['Feature Streams'] > artist_streams['Total Streams']
print(f'Artists with Lead > Total:    {lead_too_big.sum()}')
print(f'Artists with Feature > Total: {feat_too_big.sum()}')

artist_streams.loc[lead_too_big | feat_too_big, ['Artist Name', 'Total Streams', 'Lead Streams', 'Feature Streams']].head()
```

```python
rel_gap = (artist_streams['Total Streams'] - artist_streams['Lead Streams'] - artist_streams['Feature Streams']).abs() / artist_streams['Total Streams']
print(f'Rows where |Total - Lead - Feature| is above 1% of Total: {(rel_gap > 0.01).sum()}')
```
